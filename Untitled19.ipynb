```python
import pandas as pd
import os
import sys
import tkinter as tk

def copy_to_clipboard(text: str):
    """
    Копирует переданный текст в буфер обмена операционной системы.
    Использует скрытую библиотеку tkinter, не требующую установки дополнительных зависимостей.
    """
    try:
        # Создаем скрытое корневое окно tkinter
        root = tk.Tk()
        root.withdraw()  # Скрываем главное окно
        
        # Очищаем буфер на всякий случай и копируем новый текст
        root.clipboard_clear()
        root.clipboard_append(text)
        
        # Обновляем состояние окна, чтобы данные точно попали в буфер,
        # иначе они могут исчезнуть при закрытии программы
        root.update() 
        
        print(f"✅ Телефон '{text}' скопирован в буфер обмена.")
    except Exception as e:
        print(f"❌ Не удалось скопировать телефон в буфер обмена: {e}")
    finally:
        # Гарантированно уничтожаем скрытое окно после использования
        if 'root' in locals():
            try:
                root.destroy()
            except tk.TclError:
                pass

def load_data(filepath: str) -> pd.DataFrame | None:
    """Загружает Excel-файл и нормализует названия колонок."""
    if not os.path.exists(filepath):
        print(f"❌ Файл '{filepath}' не найден. Положите его в папку со скриптом.")
        return None
    
    try:
        df = pd.read_excel(filepath)
        
        # Нормализация заголовков: приводим к нижнему регистру и убираем лишние пробелы
        original_columns = df.columns.tolist()
        df.columns = [col.strip().lower() for col in df.columns]
        
        required_cols = ['фио', 'телефон', 'город']
        missing_cols = [col for col in required_cols if col not in df.columns]
        
        if missing_cols:
            print("❌ В файле найдены столбцы:")
            print(", ".join(original_columns))
            print(f"\nНо отсутствуют обязательные столбцы: {', '.join(missing_cols)}")
            print("Переименуйте их в Excel так, чтобы были: ФИО, Телефон, Город.")
            return None
            
        print(f"✅ Данные загружены успешно. Найдено кандидатов: {len(df)}")
        return df
        
    except Exception as e:
        print(f"❌ Ошибка при чтении файла Excel: {e}")
        return None

def main():
    filepath = 'candidates.xlsx'
    df = load_data(filepath)
    
    if df is None:
        sys.exit(1)

    print("\nВведите команду. Доступно:\n- город <Название_Города>\n- фамилия <Фамилия>\n- выход\n")

    while True:
        command = input(">>> ").strip().lower()

        if command == "выход":
            print("👋 Работа завершена.")
            break
        
        elif command.startswith("город"):
            parts = command.split(maxsplit=1)
            if len(parts) < 2 or not parts[1].strip():
                print("ℹ️ Используйте формат: город Москва")
                continue
                
            city_name = parts[1].strip().title()
            results = df[df['город'].str.lower() == city_name.lower()]
            
            if results.empty:
                print(f"🔍 В городе '{city_name}' кандидаты не найдены.")
            else:
                display_df = results[['фио', 'телефон', 'город', 'зарплата']]
                # to_string(index=False) выводит красивую таблицу без лишних индексов
                print(display_df.to_string(index=False))

        elif command.startswith("фамилия"):
            parts = command.split(maxsplit=1)
            if len(parts) < 2 or not parts[1].strip():
                print("ℹ️ Используйте формат: фамилия Иванов")
                continue
                
            surname = parts[1].strip().lower()
            mask = df['фио'].str.lower().str.contains(surname)
            results = df[mask]
            
            if results.empty:
                print(f"🔍 Кандидат с фамилией '{surname.title()}' не найден.")
            else:
                first_match = results.iloc[0]
                full_name = first_match['фио']
                phone = first_match['телефон']
                
                print(f"Найден: {full_name}")
                print(f"Телефон: {phone}")
                
                # Пытаемся скопировать только если номер телефона не пустой
                if pd.notna(phone) and str(phone).strip():
                    copy_to_clipboard(str(phone))
                else:
                    print("⚠️ Номер телефона пуст, копировать нечего.")

        else:
            print("Неизвестная команда. Попробуйте еще раз.")

if __name__ == "__main__":
    main()
```
